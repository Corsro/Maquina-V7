```python

#@title Region Check { display-mode: "form" }
import json, urllib.request, time
from google.colab import runtime

try:
    with urllib.request.urlopen("http://ipinfo.io/json") as resp:
        info = json.loads(resp.read().decode())
        region = info.get("region", info.get("city", "unknown"))
except Exception:
    region = "unknown"

print(f"Your current region: {region}")

try:
    print("Automatically keeping this runtime in 15 seconds... (Press Ctrl+C to unassign)")
    for i in range(15, 0, -1):
        print(f"{i}...", end="", flush=True)
        time.sleep(1)
    print("\nKeeping current runtime...")
except KeyboardInterrupt:
    print("\nCtrl+C detected — unassigning runtime...")
    runtime.unassign()
```
